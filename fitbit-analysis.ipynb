```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
activity=pd.read_csv('dailyActivity_merged.csv')
activity
```

```python
calories=pd.read_csv('dailyCalories_merged.csv')
calories
```

```python
intensity=pd.read_csv('dailyIntensities_merged.csv')
intensity
```

```python
steps=pd.read_csv('dailySteps_merged.csv')
steps
```

```python
heartrate=pd.read_csv('heartrate_seconds_merged.csv')
heartrate
```

```python
sleep=pd.read_csv('sleepDay_merged.csv')
sleep
```

```python
weight=pd.read_csv('weightlogInfo_merged.csv')
weight
```

```python
# Date column clean & same format 
activity['ActivityDate'] = pd.to_datetime(activity['ActivityDate'])
calories['ActivityDay'] = pd.to_datetime(calories['ActivityDay'])
intensity['ActivityDay'] = pd.to_datetime(intensity['ActivityDay'])
steps['ActivityDay'] = pd.to_datetime(steps['ActivityDay'])
sleep['SleepDay'] = pd.to_datetime(sleep['SleepDay'])
weight['Date'] = pd.to_datetime(weight['Date'])
heartrate['Time'] = pd.to_datetime(heartrate['Time'])
```

```python
# change heartrate 
heartrate['Date'] = heartrate['Time'].dt.date

daily_hr = heartrate.groupby(['Id', 'Date'])['Value'].mean().reset_index()
daily_hr.rename(columns={'Value': 'AvgHeartRate'}, inplace=True)

daily_hr['Date'] = pd.to_datetime(daily_hr['Date'])
```

```python
# merge all dataset
merged_df = (
    activity
    .merge(
        calories,
        left_on=["Id", "ActivityDate"],
        right_on=["Id", "ActivityDay"],
        how="left"
    )
    .merge(
        steps,
        on=["Id", "ActivityDay"],
        how="left"
    )
    .merge(
        intensity,
        on=["Id", "ActivityDay"],
        how="left"
    )
    .merge(
        sleep,
        left_on=["Id", "ActivityDate"],
        right_on=["Id", "SleepDay"],
        how="left"
    )
    .merge(
        weight,
        left_on=["Id", "ActivityDate"],
        right_on=["Id", "Date"],
        how="left"
    )
    .merge(
        daily_hr,
        left_on=["Id", "ActivityDate"],
        right_on=["Id", "Date"],
        how="left"
    )
)
```

```python
merged_df.head()
```

```python
# 1. How does step count distribution vary across users?
plt.figure(figsize=(10,5))
sns.boxplot(data=merged_df, x='TotalSteps')
plt.title('Distribution of Daily Steps Across All Users')
plt.show()

# Insight: Reveals outliers, skewness, and variability.
# Shows you understand distribution analysis, not just averages.
```

```python
# 2. What is the relationship between steps and calories burned?
plt.figure(figsize=(8,6))

sns.scatterplot(
    data=merged_df,
    x='TotalSteps',
    y='Calories_x',
    hue='Calories_x',     # color varies by value
    palette='viridis',
    alpha=0.6,
    legend=False
)

plt.title('Steps vs Calories Burned')
plt.xlabel('Total Steps')
plt.ylabel('Calories Burned')
plt.show()

# Insight: Scatter shows non-linear patterns that summary statistics hide.
```

```python
# 3. How do activity intensity minutes compare within a day?
daily_intensity = merged_df.groupby('Date_y')[
    ['LightlyActiveMinutes_x','FairlyActiveMinutes_x','VeryActiveMinutes_x']
].mean()

daily_intensity.plot(kind='bar', stacked=True, figsize=(12,5))
plt.title('Average Intensity Distribution per Day')
plt.show()

# Insight: Stacked bars reveal composition, not just totals.
```

```python
# 4. How does average sleep duration change over time?
sleep_trend = merged_df.groupby('Date_y')['TotalMinutesAsleep'].mean()

plt.figure(figsize=(12,4))
plt.plot(sleep_trend)
plt.title('Average Sleep Duration Over Time')
plt.xticks(rotation=55)
plt.show()

# Insight: Time-series visualization helps detect seasonality and trends.
```

```python
# 5. Which users contribute most to total steps?
user_steps = merged_df.groupby('Id')['TotalSteps'].sum().sort_values()

plt.figure(figsize=(10,4))
user_steps.plot(kind='bar')
plt.title('Total Steps per User')
plt.show()

# Insight: Horizontal bars improve readability for categorical comparisons.
```

```python
# 6. How concentrated are calories burned across users?
cal_sorted = merged_df.groupby('Id')['Calories_x'].sum().sort_values()
cum_cal = cal_sorted.cumsum() / cal_sorted.sum()

plt.figure(figsize=(6,6))
plt.plot(cum_cal.values)
plt.plot([0,1],[0,1],'--')
plt.title('Lorenz Curve of Calories Burned')
plt.show()

# Insight: Shows inequality in contribution, a very advanced visualization concept.
```

```python
# 7. How does sedentary time relate to active time?

plt.figure(figsize=(8,5))

hb = plt.hexbin(
    merged_df['SedentaryMinutes_x'],
    merged_df['VeryActiveMinutes_x'],
    gridsize=25,
    cmap='viridis',      # best balanced colormap
    bins='log',          # makes light colors visible
    edgecolors='grey',
    linewidths=0.2
)

plt.colorbar(hb, label='Density')

plt.xlabel('Sedentary Minutes per Day', fontsize=11)
plt.ylabel('Very Active Minutes per Day', fontsize=11)

plt.title('Sedentary Time vs Very Active Time', fontsize=13)

plt.tight_layout()
plt.show()

# Insight: Hexbin handles overplotting better than scatter for large data.
```
