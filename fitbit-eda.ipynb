```python
# Date column clean & same format 
activity['ActivityDate'] = pd.to_datetime(activity['ActivityDate'])
calories['ActivityDay'] = pd.to_datetime(calories['ActivityDay'])
intensity['ActivityDay'] = pd.to_datetime(intensity['ActivityDay'])
steps['ActivityDay'] = pd.to_datetime(steps['ActivityDay'])
sleep['SleepDay'] = pd.to_datetime(sleep['SleepDay'])
weight['Date'] = pd.to_datetime(weight['Date'])
heartrate['Time'] = pd.to_datetime(heartrate['Time'])
```

```python
# change heartrate 
heartrate['Date'] = heartrate['Time'].dt.date

daily_hr = heartrate.groupby(['Id', 'Date'])['Value'].mean().reset_index()
daily_hr.rename(columns={'Value': 'AvgHeartRate'}, inplace=True)

daily_hr['Date'] = pd.to_datetime(daily_hr['Date'])
```

```python
# merge all dataset
merged_df = (
    activity
    .merge(
        calories,
        left_on=["Id", "ActivityDate"],
        right_on=["Id", "ActivityDay"],
        how="left"
    )
    .merge(
        steps,
        on=["Id", "ActivityDay"],
        how="left"
    )
    .merge(
        intensity,
        on=["Id", "ActivityDay"],
        how="left"
    )
    .merge(
        sleep,
        left_on=["Id", "ActivityDate"],
        right_on=["Id", "SleepDay"],
        how="left"
    )
    .merge(
        weight,
        left_on=["Id", "ActivityDate"],
        right_on=["Id", "Date"],
        how="left"
    )
    .merge(
        daily_hr,
        left_on=["Id", "ActivityDate"],
        right_on=["Id", "Date"],
        how="left"
    )
)
```

```python
merged_df.head()
```

```python
# 1. First use the daily activity data and find the unique id of all the users.
activity['Id'].unique()
```

```python
# 2. Then find the number of days they use their fitness tracker.
user_days = merged_df.groupby('Id')['ActivityDate'].nunique().reset_index()
user_days.columns = ['Id', 'TotalDaysUsed']

user_days.head()
```

```python
# 3. Use this data to group all the users into Active(>20), moderate(>10,<20) and light(<10) users.
def user_type(days):
    if days > 20:
        return 'Active'
    elif days >= 10:
        return 'Moderate'
    else:
        return 'Light'

user_days['UserType'] = user_days['TotalDaysUsed'].apply(user_type)
user_days.head()
```

```python
# 4. After that find out the mean distance traveled by each user
mean_distance=merged_df.groupby('Id')['TotalDistance'].mean().reset_index()
mean_distance.columns=['Id','MeanDistance']
mean_distance.head()
```

```python
# 5. Use this mean distance to group each user into Pro,Intermediate and beginner.(set criteria by yourself)
def distance_level(dist):
    if dist>10:
        return 'Pro'
    elif dist>=5:
        return 'Intermediate'
    else:
        return 'Beginner'
mean_distance['Level']=mean_distance['MeanDistance'].apply(distance_level)
mean_distance.head()
```

```python
# 6. Then find out the total steps traveled by each user
total_steps=merged_df.groupby('Id')['TotalSteps'].sum().reset_index()
total_steps.columns=['ID','TotalSteps']
total_steps.head()
```

```python
# 7 -> Total calories burn per user
total_calories=merged_df.groupby('Id')['Calories_x'].sum().reset_index()
total_calories.columns=['ID','TotalCalories']
total_calories.head()
```

```python
# 8 -> Active minutes per user
activity_minutes = merged_df.groupby('Id')[[
    'VeryActiveMinutes_x',
    'FairlyActiveMinutes_x',
    'LightlyActiveMinutes_x'
]].sum().reset_index()

activity_minutes.head()

```

```python
# 9. Repeat all these tasks (1 to 8) with respect to distinct dates
# (a) Unique users per date
user_per_date=merged_df.groupby('ActivityDate')['Id'].nunique().reset_index()
user_per_date.columns=['Date','TotalUsers']
user_per_date.head()
```

```python
# (b) Steps per date
steps_per_date=merged_df.groupby('ActivityDate')['TotalSteps'].nunique().reset_index()
steps_per_date.columns=['Date','TotalSteps']
steps_per_date.head()
```

```python
# (c) Calories per date
calories_per_date=merged_df.groupby('ActivityDate')['Calories_x'].sum().reset_index()
calories_per_date.head()
```

```python
# (d) Activity minutes per date
minutes_per_date = merged_df.groupby('ActivityDate')[[
    'VeryActiveMinutes_x',
    'FairlyActiveMinutes_x',
    'LightlyActiveMinutes_x'
]].sum().reset_index()

minutes_per_date.head()

```

```python
# 32. Can we create a user-level “Activity Stability Index” using a custom function, and identify the most stable users?

def activity_stability(group):
    step_std = group['TotalSteps'].std()
    active_std = group['VeryActiveMinutes_x'].std()
    
    # lower std = higher stability
    stability_score = 1 / (step_std + active_std + 1)
    return stability_score

stability_index = (
    merged_df
    .groupby('Id')
    .apply(activity_stability)
    .sort_values(ascending=False)
)

stability_index.head(10).reset_index()


# Insight:  This index captures behavioral consistency, not performance. High stability users are ideal for long-term habit modeling, even if their absolute activity is moderate.
```

```python
# 33. Can we classify daily behavior into intensity profiles using a custom mapping function?
def activity_profile(row):
    if row['VeryActiveMinutes_x'] > row['LightlyActiveMinutes_x']:
        return 'Intensity-Driven'
    elif row['LightlyActiveMinutes_x'] > row['SedentaryMinutes_x']:
        return 'Light-Active'
    else:
        return 'Sedentary-Dominant'

merged_df['activity_profile'] = merged_df.apply(activity_profile, axis=1)

profile_distribution = (
    merged_df
    .groupby('Id')['activity_profile']
    .value_counts(normalize=True)
)

profile_distribution.head(10).reset_index()

# Insight : This shows how users spend their energy, not how much. Such categorical profiling is highly reusable in customer behavior analytics.
```

```python
# 34. Can we build a custom “Recovery Efficiency Score” using sleep and next-day activity with a lambda function?
def recovery_efficiency(group):
    group = group.sort_values('Date_y')
    
    next_day_steps = group['TotalSteps'].shift(-1)
    sleep = group['TotalMinutesAsleep']
    
    efficiency = (next_day_steps / (sleep + 1)).mean()
    return efficiency

recovery_score = (
    merged_df
    .groupby('Id')
    .apply(recovery_efficiency)
    .sort_values(ascending=False)
)

recovery_score.head(10).reset_index()

# Insight: This captures sleep quality impact, not sleep duration alone. Lag-based feature engineering = advanced EDA skill.
```

```python
# 35. Can we quantify how sensitive each user’s calories are to changes in intensity using a custom regression-like function?
def intensity_sensitivity(group):
    if group['VeryActiveMinutes_x'].std() == 0:
        return np.nan
    
    sensitivity = (
        group['Calories_x'].cov(group['VeryActiveMinutes_x']) /
        group['VeryActiveMinutes_x'].var()
    )
    return sensitivity

intensity_effect = (
    merged_df
    .groupby('Id')
    .apply(intensity_sensitivity)
    .sort_values(ascending=False)
)

intensity_effect.head(10).reset_index()

# Insight: This approximates marginal calorie response without ML.Shows you understand relationships, not just correlations.
```

```python
# 37. Can we detect users whose behavior changes drastically over time using a rolling-window function?
def behavior_shift(group):
    group = group.sort_values('Date_x')
    
    early = group['TotalSteps'].rolling(7).mean().iloc[:7].mean()
    late = group['TotalSteps'].rolling(7).mean().iloc[-7:].mean()
    
    return late - early

behavior_change = (
    merged_df
    .groupby('Id')
    .apply(behavior_shift)
    .sort_values()
)

behavior_change.head(10).reset_index()

# Insight: This detects lifestyle change, not noise.Same logic applies to churn detection in any industry.
```

```python
# 38. Can we create a composite “Daily Load Score” using weighted mapping of multiple activity metrics?
def daily_load(row):
    return (
        row['VeryActiveMinutes_x'] * 3 +
        row['FairlyActiveMinutes_x'] * 2 +
        row['LightlyActiveMinutes_x'] * 1
    )

merged_df['daily_load'] = merged_df.apply(daily_load, axis=1)

user_load = (
    merged_df
    .groupby('Id')['daily_load']
    .mean()
    .sort_values(ascending=False)
)

user_load.head(10).reset_index()

# Insight: Weighted feature construction shows domain understanding,not blind aggregation.
```

```python
# 39. Can we identify users whose activity is driven more by routine than motivation using a custom variance ratio?
def routine_strength(group):
    weekday_means = (
        group
        .groupby(group['Date_x'].dt.day_name())['TotalSteps']
        .mean()
    )
    return 1 / (weekday_means.std() + 1)

routine_index = (
    merged_df
    .groupby('Id')
    .apply(routine_strength)
    .sort_values(ascending=False)
)

routine_index.head(10).reset_index()

# Insight: This measures structural discipline, not effort.Very strong behavioral analytics signal.
```

```python
# 40. Can we design a “Sustainability Score” using a custom function that penalizes over-training and poor sleep?
def sustainability_score(group):
    avg_intensity = group['VeryActiveMinutes_x'].mean()
    avg_sleep = group['TotalMinutesAsleep'].mean()
    sedentary_ratio = group['SedentaryMinutes_x'].mean() / 1440
    
    score = (
        avg_sleep * 0.4 +
        (1 - sedentary_ratio) * 0.3 -
        avg_intensity * 0.01
    )
    return score

sustainability = (
    merged_df
    .groupby('Id')
    .apply(sustainability_score)
    .sort_values(ascending=False)
)

sustainability.head(10).reset_index()

# Insight: This reflects long-term health sustainability, not short-term performance.Composite scoring like this is senior-level EDA.
```
